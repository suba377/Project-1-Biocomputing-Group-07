##### 2(a) Using the CDS coordinates from 1(b), extract the coding sequence from the mRNA. Report its length in nucleotides and confirm that it is divisible by three. If it is not, stop and work out why before continuing — do not proceed with a sequence that fails this check.

```python
import Bio
```

```python
from Bio import SeqIO

"""Reading the Human Mitochondrial Genome stored in fasta file and store in record"""
record = SeqIO.read("MT-ND4_Mitochondrial_Gene.fasta", "fasta")

""""Extracting gene sequence of MT-ND4 from the mitochondrial genome using CDS coordinates"""
start_codon = 10760
stop_codon = 12137
coding_seq = record.seq[start_codon-1:stop_codon] 

""""Transcribing coding sequence to get mRNA sequence"""
mrna = coding_seq.transcribe()

print("Length of mrna:", len(mrna))
print("Is it divisible by 3:", len(mrna)/3 == 0)
print("Remainder left:", len(mrna)%3)
```

```python
print(mrna)
```

The 1 remainder i.e. end of the mRNA sequence consists of U. It is the stop codon It is noted in the feature table of the MT-ND4 gene that stop codon is completed by adding 3’ A residues. So, we can add A, A residues at the end to complete the stop codon.

```python
"""Add A, A residues at the end to complete the stop codon."""
mrna_AA = mrna + "AA"

print("New length of the mRNA:", len(mrna_AA))
print("Is it divisible by 3:", len(mrna_AA)/3 == 0)
print("Remainder left:", len(mrna_AA)%3)
```

##### 2(b)Build the genetic code as a lookup structure of your own making, then use it to translate the entire coding sequence to protein. Report the first 30 residues of your translation and total length.You must construct the lookup yourself. You may read the codon assignments from a published table, but the data structure and the translation loop must be your own work.

```python
""""Creating mitochondrial genetic codon table"""
codon_table = {
    "UUU": "F","AUG": "M","UCU": "S","ACU": "T","AGU": "S","CAU": "H",
    "UUC": "F","GUU": "V","UCC": "S","ACC": "T","AGC": "S","CAC": "H",
    "UUA": "L","GUC": "V","UCA": "S","ACA": "T","GAG": "E","CAA": "Q",
    "UUG": "L","GUA": "V","UCG": "S","ACG": "T","GAA": "E","CAG": "Q",
    "CUU": "L","GUG": "V","CCU": "P","GCU": "A","GAU": "D","UGU": "C",
    "CUC": "L","CGU": "R","CCC": "P","GCC": "A","GAC": "D","UGC": "C",
    "CUA": "L","CGC": "R","CCA": "P","GCA": "A","AAU": "N","UGG": "W",
    "CUG": "L","CGA": "R","CCG": "P","GCG": "A","AAC": "N","UAA": "*",
    "AUU": "I","CGG": "R","UAU": "Y","GGC": "G","AAA": "K","UAG": "*",
    "AUC": "I","AGA": "*","UAC": "Y","GGA": "G","AAG": "K","UGA": "W",
    "AUA": "M","AGG": "*","GGU": "G","GGG": "G"
}
```

```python
""""Translating mRNA sequence into protein using codon table"""
protein = ""
for i in range(0, len(mrna_AA), 3):
    codon = mrna_AA[i:i+3]
    amino_acid = codon_table[codon]
    protein += amino_acid

print("First 30 residues of translated protein are:", protein[:30])
print("Total length of the translated protein is:", len(protein))

"""Removing the stop codon symbol from protein sequence as it's actually not part of the protein"""
protein = protein[:459]
```
