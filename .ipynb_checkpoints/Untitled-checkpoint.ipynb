2(a) (2 points)
Using the CDS coordinates from 1(b), extract the coding sequence from the mRNA. Report its length in
nucleotides and confirm that it is divisible by three. If it is not, stop and work out why before continuing —
do not proceed with a sequence that fails this check.

to get mrna



```python
import Bio
```

```python
from Bio import SeqIO

record = SeqIO.read("MT-ND4_Mitochondrial_Gene.fasta", "fasta")

start_codon = 10760
stop_codon = 12137
coding_seq = record.seq[start_codon-1:stop_codon]
mrna = coding_seq.transcribe()

print("Length of mrna:", len(mrna))
print("Dividing it by 3:", len(mrna)/3)
print("Remainder left:", len(mrna)%3)
print(mrna)

mrna_AA = mrna + "AA"

print(mrna_AA)
print("New length:", len(mrna_AA))
print("Dividing it by 3:", len(mrna_AA)/3)
print("Remainder left:", len(mrna_AA)%3)
```

it's mentioned that TAA stop codon is incomplete and to the mRNA 3' A residues are added

2(b) (4 points)
Build the genetic code as a lookup structure of your own making, then use it to translate the entire coding sequence to protein. Report the first 30 residues of your translation and total length.
You must construct the lookup yourself. You may read the codon assignments from a published table, but
the data structure and the translation loop must be your own work.

```python
codon_table = {
    "UUU": "F","AUG": "M","UCU": "S","ACU": "T","AGU": "S","CAU": "H",
    "UUC": "F","GUU": "V","UCC": "S","ACC": "T","AGC": "S","CAC": "H",
    "UUA": "L","GUC": "V","UCA": "S","ACA": "T","GAG": "E","CAA": "Q",
    "UUG": "L","GUA": "V","UCG": "S","ACG": "T","GAA": "E","CAG": "Q",
    "CUU": "L","GUG": "V","CCU": "P","GCU": "A","GAU": "D","UGU": "C",
    "CUC": "L","CGU": "R","CCC": "P","GCC": "A","GAC": "D","UGC": "C",
    "CUA": "L","CGC": "R","CCA": "P","GCA": "A","AAU": "N","UGG": "W",
    "CUG": "L","CGA": "R","CCG": "P","GCG": "A","AAC": "N","UAA": "*",
    "AUU": "I","CGG": "R","UAU": "Y","GGC": "G","AAA": "K","UAG": "*",
    "AUC": "I","AGA": "*","UAC": "Y","GGA": "G","AAG": "K","UGA": "W",
    "AUA": "M","AGG": "*","GGU": "G","GGG": "G"
}

```

```python
protein = ""
for i in range(0, len(mrna_AA), 3):
    codon = mrna_AA[i:i+3]
    amino_acid = codon_table[codon]
    protein += amino_acid

print(protein)
print("First 30 residues:", protein[:30])
print("Total length:", len(protein))
```
